```python
import os
from dotenv import load_dotenv

load_dotenv()

from openai import OpenAI

# 本课改用国内 OpenAI 兼容 API（月之暗面 Kimi），走标准 Chat Completions 接口。
# 原因：中国大陆个人 Azure 账号受 CannotDeployDueToLocalRegulations 政策限制，
# 无法部署 Azure OpenAI / Foundry 模型，也没有 Entra ID 凭据可用。
api_key = os.environ["LLM_API_KEY"]
base_url = os.environ["LLM_BASE_URL"]
model = os.environ["LLM_MODEL"]
```

```python
# 国内 API 用静态 api_key 鉴权，无需 Entra ID 凭据。
client = OpenAI(
    api_key=api_key,
    base_url=base_url,
)
```

```python
role = "旅行代理"
company = "远景旅行"
responsibility = "预订机票"
```

```python
response = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "system", "content": """你是一位设计 AI 智能体助手的专家。
你会拿到公司名称、角色、职责等信息，需要据此为这个智能体写出一份系统提示词（system prompt）。
请尽可能详尽，并采用结构化表达，让使用大模型的系统能清楚理解该 AI 助手的角色与职责边界。
请用中文输出。"""},
        {"role": "user", "content": f"你是 {company} 的{role}，负责{responsibility}。"},
    ],
    # kimi-k2.6 为思考型模型，temperature 固定为 1.0
    temperature=1.0,
    max_tokens=1000,
    top_p=1.0,
)

print(response.choices[0].message.content)
```
